
# Udemy Course Data Analysis (Learning Project)

This project focuses on analyzing Udemy course data across **four subject categories**:
**Business Finance, Graphic Design, Musical Instruments, and Web Development**.

The goal of this analysis is to understand how courses perform across different subjects by examining the **number of courses**, **subscriber trends**, **course ratings**, and **revenue generation**. In addition, the project explores **common keywords** associated with highly rated and highly subscribed courses within each subject area.



### Key Questions Explored

This analysis aims to answer the following questions:

* Which subject area has the **highest and lowest number of published courses**?
* Which subject area attracts the **largest number of subscribers**?
* Which **course levels** are offered as free courses?
* What **keywords** frequently appear in courses with high ratings and high subscription counts?
* How do **course ratings** vary across different subject areas?
* For each subject category, **which year saw the highest number of course publications**?
* Which subject area contributes the **most to overall revenue**?

### Data Wrangling

```python
#check data types and missing values
df.info()
```

```python
#check for and drop duplicates
df[df.duplicated()]

#drop duplicates in the course id column
df.drop_duplicates(subset=['course_id'],inplace=True)
```

```python
#drop rows with missing values
df.dropna(inplace=True)

#confirm there is no more any null values
df.isnull().sum()
```

We wouldn't be needing the columns below for our analysis; hence we shall drop them. We will change the column data types to the most suitable.

```python
#drop unwanted columns from the combined dataframe
df = df.drop(['url','num_reviews', 'course_id', 'num_lectures','content_duration'], axis = 1)
```

```python
#cast datatype as int
df['num_subscribers'] = df['num_subscribers'].astype('int64')


#confirm change
df.head(2)
```

```python
import numpy as np

conditions = [
    df['price'] == 0,
    df['price'] > 0
]

values = ['free', 'paid']

df['price_group'] = np.select(conditions, values, default='unknown')

df['price_group'].value_counts()
```

Below we will group the ratings into bins to allow for some easy referencing.

```python
#set category for ratings using a 5star system
conditions1 = [
    (df['Rating'] == 0),
    (df['Rating'] > 0) & (df['Rating'] < 0.2),
    (df['Rating'] >= 0.2) & (df['Rating'] < 0.4),
    (df['Rating'] >= 0.4) & (df['Rating'] < 0.6),
    (df['Rating'] >= 0.6) & (df['Rating'] < 0.8),
    (df['Rating'] >= 0.8)
]
values1 = [0, 1, 2, 3, 4, 5]
df['star_rating'] = np.select(conditions1, values1)

df['star_rating'].value_counts()
```

```python
#rename the levels
levels = ({'All Levels':'General', 'Beginner Level':'Beginner',
           'Intermediate Level':'Intermediate', 'Expert Level':'Expert'})
df['level'] = df['level'].replace(levels)

df['level'].value_counts()
```

Let us extract the date only object from the datetime supplied; we wont be needing the time object for this analysis. We will then create the `year` column.

```python
#extract date only from datetime object
df['published'] = df['published_timestamp'].str.split('T').str[0]

df = df.drop(['published_timestamp'], axis = 1) #drop the combined timestam column

df['published']
```

```python
#create column for the year the course was published
df['year'] = (df['published'].str.split('-').str[0]).astype(int)
df['year'].value_counts()
```

Let us clean the text columns a bit.

```python
#write regex funtion to clean text in course_title column

def clean(text):
    text = str(text).title()
    text = re.sub('\[.*?\]', '', text)
    text = re.sub('https?://\S+|www\.\S+', '', text) #remove url
    text = re.sub('[%s]' % re.escape(string.punctuation), '', text) #remove punctuations
    text = re.sub('\n', '', text)
    text = re.sub('[0-9]', '', text)
    text = re.sub('<.*?>+', '', text)
    return text

df['course_title'] = df['course_title'].apply(clean)
```

```python
#clean off punctautions from subject column

df['subject']=df['subject'].str.split(': ').str[-1].str.lstrip()
df['subject'].value_counts()
```

Let us create the `Revenue` column by multiplying the `num_subscribers` by `price` of the course.

```python
df['revenue'] = df['num_subscribers'] * df['price']
```

```python
#separate the component dataframes

df_biz = df.query('subject == "Business Finance"')
df_gfx = df.query('subject == "Graphic Design"')
df_mus = df.query('subject == "Musical Instruments"')
df_dev = df.query('subject == "Web Development"')

print('There are {} {} courses'.format(df_biz.shape[0], df_biz.subject[0]))
print('There are {} {} courses'.format(df_gfx.shape[0], df_gfx.subject[0]))
print('There are {} {} courses'.format(df_mus.shape[0], df_mus.subject[0]))
print('There are {} {} courses'.format(df_dev.shape[0], df_dev.subject[0]))
```

Let us see the common words used to frame courses that have above 4-star rating and above average number of subscription across the different subjects.

```python

# 1. IMPORTS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from wordcloud import WordCloud, STOPWORDS


# 2. DOWNLOAD REQUIRED NLTK DATA
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('punkt')


# 3. INITIALIZE NLP TOOLS

wn = WordNetLemmatizer()
stopword = set(stopwords.words('english'))

# =========================
# 4. TEXT CLEANING FUNCTION
# =========================
def word_join(text):
    text = str(text).lower()
    text = ' '.join([word for word in text.split() if word not in stopword])
    text = ' '.join([wn.lemmatize(word, pos='v') for word in text.split()])
    return text


# 5. WORDCLOUD FOR ALL COURSES

text = ' '.join(df['course_title'].apply(word_join))

wordcloud = WordCloud(
    stopwords=STOPWORDS,
    background_color='black',
    max_words=50
).generate(text)

plt.figure(figsize=(15,10))
plt.imshow(wordcloud, interpolation='bilinear')
plt.title('Key Words in All Courses')
plt.axis('off')
plt.show()


# 6. WORDCLOUDS FOR TOP RATED COURSES BY CATEGORY

fig, ((ax0, ax1), (ax2, ax3)) = plt.subplots(2, 2, figsize=(15,10))

# ---- Business Finance ----
df_bizt = df_biz.query('star_rating >= 4 & num_subscribers > 1569')
text = ' '.join(df_bizt['course_title'].apply(word_join))
wc = WordCloud(background_color='black', max_words=50).generate(text)
ax0.imshow(wc)
ax0.set_title('Top Business Finance Keywords')
ax0.axis('off')

# ---- Graphic Design ----
df_gfxt = df_gfx.query('star_rating >= 4 & num_subscribers > 1766')
text = ' '.join(df_gfxt['course_title'].apply(word_join))
wc = WordCloud(background_color='black', max_words=50).generate(text)
ax1.imshow(wc)
ax1.set_title('Top Graphic Design Keywords')
ax1.axis('off')

# ---- Music ----
df_must = df_mus.query('star_rating >= 4 & num_subscribers > 1245')
text = ' '.join(df_must['course_title'].apply(word_join))
wc = WordCloud(background_color='black', max_words=50).generate(text)
ax2.imshow(wc)
ax2.set_title('Top Music Course Keywords')
ax2.axis('off')

# ---- Web Development ----
df_devt = df_dev.query('star_rating >= 4 & num_subscribers > 6635')
text = ' '.join(df_devt['course_title'].apply(word_join))
wc = WordCloud(background_color='black', max_words=50).generate(text)
ax3.imshow(wc)
ax3.set_title('Top Web Development Keywords')
ax3.axis('off')

plt.tight_layout()
warnings.filterwarnings("ignore")
plt.show()
```
